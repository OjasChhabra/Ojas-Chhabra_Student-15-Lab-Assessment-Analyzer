## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
# TODO: Select numeric columns relevant to your project
marks1 = df["Experiment_1"]
marks2 = df["Experiment_2"] 
marks3 = df["Experiment_3"] 
viva = df[" Viva"]
# TODO: Convert them to a NumPy array using .to_numpy()
list1 = marks1.to_numpy()
list2 = marks2.to_numpy()
list3 = marks3.to_numpy()
list4 = viva.to_numpy()
# TODO: Calculate average, maximum and minimum
print("Experiment 1: ")
print("Average:", np.mean(list1))
print("Highest:", np.max(list1))
print("Lowest:", np.min(list1),"\n")
print("Experiment 2: ")
print("Average:", np.mean(list2))
print("Highest:", np.max(list2))
print("Lowest:", np.min(list2),"\n")
print("Experiment 3: ")
print("Average:", np.mean(list3))
print("Highest:", np.max(list3))
print("Lowest:", np.min(list3),"\n")
print("Viva: ")
print("Average:", np.mean(list4))
print("Highest:", np.max(list4))
print("Lowest:", np.min(list4),"\n")
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# TODO: Create calculated column(s)
# TODO: Filter records using a condition
# TODO: Sort records and display the top/bottom results
df["Total_Marks"] = df["Experiment_1"] + df["Experiment_2"] + df["Experiment_3"] + df[" Viva"]
display(df)
```

## 6. Your individual project challenge
**Challenge:** Calculate practical average; identify students below 40 in any component; rank students.

Write your solution below and explain what the result means.


```python
# TODO: Solve your assigned challenge
df["Average_Marks"] = df["Total_Marks"] / 4
df["Any_Marks_below_40"] = (
    (df["Experiment_1"] < 40) |
    (df["Experiment_2"] < 40) |
    (df["Experiment_3"] < 40) |
    (df[" Viva"] < 40)
)
df["Rank"] = df["Average_Marks"].rank(ascending=False)
df = df.sort_values(by="Rank")
display(df)
```
